```python
import os
import numpy as np
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Garantir que a variável caminho existe, caso contrário usa o diretório padrão /content/
if 'caminho' not in globals():
    caminho = '/content/drive/My Drive/FACULDADE/3 Semestre/API/'

# Caso o notebook tenha sido resetado e 'Dados' não esteja na memória, vamos recarregá-lo automaticamente
if 'Dados' not in globals():
    print("Variável 'Dados' não encontrada. Recarregando planilha Dados_Tratados.csv...")
    try:
        Dados = pd.read_csv(caminho + 'Dados_Tratados.csv')
        Dados['ENDERECO_COMPLETO'] = Dados['LOGRADOURO'].astype(str) + ', ' + Dados['CIDADE'].astype(str) + ' - SP, Brasil'
        print("Dados carregados com sucesso!")
    except Exception as e:
        print(f"Não foi possível carregar os dados automaticamente: {e}")
        print("Por favor, certifique-se de executar as primeiras células que montam o Google Drive primeiro.")

# Caminho para salvar o progresso e não perder dados caso caia a conexão
arquivo_progresso = caminho + 'enderecos_geocodificados_parcial.csv'

# Configurando o geocodificador com timeout de 10 segundos
geolocator = Nominatim(user_agent="meu_app_api_rotas_v3", timeout=10)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1.2, swallow_exceptions=True)

# Carregar progresso anterior se existir, caso contrário criar um novo DataFrame
if os.path.exists(arquivo_progresso):
    print("Encontrado progresso anterior! Carregando arquivo parcial...")
    enderecos_unicos = pd.read_csv(arquivo_progresso)
else:
    print("Criando novo controle de endereços únicos...")
    enderecos_unicos = pd.DataFrame(Dados['ENDERECO_COMPLETO'].unique(), columns=['ENDERECO_COMPLETO'])
    enderecos_unicos['LATITUDE'] = np.nan
    enderecos_unicos['LONGITUDE'] = np.nan
    enderecos_unicos['PROCESSADO'] = False

# Filtrar apenas os que ainda não foram processados com sucesso
pendentes = enderecos_unicos[enderecos_unicos['PROCESSADO'] == False].copy()
total_pendentes = len(pendentes)

print(f"Total de endereços únicos: {len(enderecos_unicos)}")
print(f"Endereços restantes para geolocalizar: {total_pendentes}")

# Configurações do processamento em lotes (Lotes de 50 para salvar frequentemente)
tamanho_lote = 50
contador = 0

if total_pendentes > 0:
    print("Iniciando busca em lotes parciais...")

    for i in range(0, total_pendentes, tamanho_lote):
        sub_lote = pendentes.iloc[i:i+tamanho_lote].copy()
        print(f"\nProcessando lote {i // tamanho_lote + 1} ({len(sub_lote)} endereços)... ")

        # Aplicar geolocalização no lote atual
        for idx, row in sub_lote.iterrows():
            endereco = row['ENDERECO_COMPLETO']
            try:
                loc = geocode(endereco)
                if loc:
                    enderecos_unicos.at[idx, 'LATITUDE'] = loc.latitude
                    enderecos_unicos.at[idx, 'LONGITUDE'] = loc.longitude
                enderecos_unicos.at[idx, 'PROCESSADO'] = True
            except Exception as e:
                # Se falhar, manter como não processado para tentar em outra execução
                print(f"Erro ao processar: {endereco}. Erro: {e}")

        # Salva o progresso atual do DataFrame inteiro no Google Drive após cada lote
        enderecos_unicos.to_csv(arquivo_progresso, index=False)
        percentual = ((len(enderecos_unicos[enderecos_unicos['PROCESSADO'] == True])) / len(enderecos_unicos)) * 100
        print(f"Lote finalizado! Progresso total salvo: {percentual:.2f}%")

# Mostrar resultado final
encontrados = enderecos_unicos['LATITUDE'].notna().sum()
print(f"\nBusca concluída! Encontrados {encontrados} de {len(enderecos_unicos)} endereços.")
display(enderecos_unicos.head())
```

```python
# 7. Carregar o arquivo final consolidado que salvamos passo a passo no drive
if 'caminho' not in globals():
    caminho = '/content/drive/My Drive/FACULDADE/3 Semestre/API/'

arquivo_progresso = caminho + 'enderecos_geocodificados_parcial.csv'

print("Carregando resultados da geolocalização por lotes...")
enderecos_geocodificados = pd.read_csv(arquivo_progresso)

# Remover a coluna de controle 'PROCESSADO' para não poluir a base final
if 'PROCESSADO' in enderecos_geocodificados.columns:
    enderecos_geocodificados = enderecos_geocodificados.drop(columns=['PROCESSADO'])

# Se a tabela Dados já continha as colunas LATITUDE e LONGITUDE de execuções parciais anteriores, vamos limpá-las antes de mesclar
for col in ['LATITUDE', 'LONGITUDE']:
    if col in Dados.columns:
        Dados = Dados.drop(columns=[col])

# Mesclar as coordenadas de volta para o DataFrame original ('Dados')
Dados = Dados.merge(enderecos_geocodificados, on='ENDERECO_COMPLETO', how='left')

print(f"Pronto! Coordenadas mescladas na tabela original. Total de linhas: {len(Dados)}")
display(Dados[['DATAVERIFICACAO', 'ENDERECO_COMPLETO', 'LATITUDE', 'LONGITUDE']].head())
```

```python
# Definir caminhos para os arquivos de saída
caminho_completo = caminho + 'Dados_Com_Coordenadas_Completo.csv'
caminho_filtrado = caminho + 'Dados_Com_Coordenadas_Apenas_Sucesso.csv'

# 1. Salvar a tabela completa (incluindo linhas sem coordenadas/NaN)
Dados.to_csv(caminho_completo, index=False)
print(f"Tabela completa salva com sucesso em:\n-> {caminho_completo}")
print(f"Total de registros: {len(Dados)} linhas.\n")

# 2. Filtrar apenas as linhas com coordenadas válidas (remover NaN de LATITUDE/LONGITUDE)
Dados_filtrados = Dados.dropna(subset=['LATITUDE', 'LONGITUDE'])

# Salvar a tabela filtrada
Dados_filtrados.to_csv(caminho_filtrado, index=False)
print(f"Tabela contendo apenas sucessos salva em:\n-> {caminho_filtrado}")
print(f"Total de registros encontrados: {len(Dados_filtrados)} linhas.")
```
